```python
# Подключение библиотек
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Настройка отображения графиков
%matplotlib inline
sns.set_style("whitegrid")
```

```python
# Загрузка данных
df = pd.read_csv('insurance.csv')

# Просмотр первых 5 строк
df.head()
```

```python
# Проверка типов данных
df.dtypes
```

```python
# Проверка пропусков в данных
df.isnull().sum()
```

```python
# Преобразование типов данных
df['age'] = df['age'].astype(int)
df['children'] = df['children'].astype(int)

# Проверка результата
df.dtypes
```

```python
# Проверка пропусков
df.isnull().sum()
```

```python
# Описательная статистика
df.describe()
```

```python
# Проверка уникальных значений в категориальных колонках
df['sex'].unique()
df['smoker'].unique()
df['region'].unique()
```

```python
# Гистограмма распределения возраста
import matplotlib.pyplot as plt
import seaborn as sns

sns.histplot(df['age'], bins=20, kde=True)
plt.title('Распределение возраста')
plt.xlabel('Возраст')
plt.ylabel('Частота')
plt.show()
```

```python
# Гистограмма распределения BMI
sns.histplot(df['bmi'], bins=20, kde=True)
plt.title('Распределение индекса массы тела (BMI)')
plt.xlabel('BMI')
plt.ylabel('Частота')
plt.show()
```

```python
# Boxplot: страховые выплаты по признаку курения
sns.boxplot(x='smoker', y='charges', data=df)
plt.title('Страховые выплаты: курящие vs некурящие')
plt.show()
```

```python
# Boxplot: страховые выплаты по полу
sns.boxplot(x='sex', y='charges', data=df)
plt.title('Страховые выплаты по полу')
plt.show()
```

```python
# Тепловая карта корреляций
plt.figure(figsize=(8, 6))
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap='coolwarm')
plt.title('Корреляционная матрица')
plt.show()
```

```python
# Средние выплаты по регионам
df.groupby('region')['charges'].mean().sort_values(ascending=False)
```
